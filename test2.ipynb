```python
import numpy as np
from elo import Elo
from glicko import Glicko
from glicko2 import Glicko2
import matplotlib.pyplot as plt
```

```python
num_players = 20
init_rating = np.full(num_players, 1500)
# init_rating = np.array([0, 0, 500, 500, 1000, 1000, 1500, 1500, 2000, 2000])
# init_true_rating = np.random.uniform(0, 3000, num_players).astype(np.float32)
# init_true_rating = np.random.rand(num_players).astype(np.float32) * 3000
# init_true_rating = np.random.normal(1500, 350, num_players).astype(np.float32)
init_true_rating = np.full(num_players, 1500)
```

```python
elo = Elo(num_players, init_rating, init_true_rating)

for _ in range(3):
    elo.match_all_realistic()
    print("Elo Ratings:", elo.ratings)

plt.figure(figsize=(10, 6))
plt.hist([elo.ratings, elo.true_ratings], bins=10, alpha=0.5, 
         label=["Elo", "True Ratings"])
plt.title("Ratings Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Players")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
